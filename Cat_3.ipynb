```python

import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report


np.random.seed(42)

data = pd.DataFrame({
    "GPA": np.random.uniform(1.5, 4.0, 300),
    "AttendanceRate": np.random.randint(50, 100, 300),
    "Age": np.random.randint(17, 30, 300),
    "Gender": np.random.randint(0, 2, 300),
    "AppliedEarly": np.random.randint(0, 2, 300),
    "FinancialAid": np.random.randint(0, 2, 300),
})

data["Enrolled"] = (
    (data["GPA"] > 2.5).astype(int)
    + (data["AttendanceRate"] > 70).astype(int)
    + data["AppliedEarly"]
    + data["FinancialAid"]
)
data["Enrolled"] = (data["Enrolled"] > 2).astype(int)  # Convert to 0/1

print("Student Enrollment Prediction:")
display(data.head())

X = data.drop("Enrolled", axis=1)
y = data["Enrolled"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

model = RandomForestClassifier(
    n_estimators=200,
    max_depth=8,
    random_state=42
)
model.fit(X_train, y_train)

y_pred = model.predict(X_test)

print("Model Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:")
print(classification_report(y_test, y_pred))

new_student = pd.DataFrame({
    "GPA": [2.7],
    "AttendanceRate": [83],
    "Age": [19],
    "Gender": [1],
    "AppliedEarly": [0],
    "FinancialAid": [1]
})

new_student_scaled = scaler.transform(new_student)
prediction = model.predict(new_student_scaled)

print("\nPrediction for New Student (1 = Enroll, 0 = Not Enroll):",
      prediction[0])
```
